```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('customer_segmentation_mixed.csv')
```

```python
df.head()
```

# 1. Categorical data

a. Countplot

```python
sns.countplot(df['Preferred_Product'])
```

b. PieChart

```python
df['Gender'].value_counts().plot(kind = 'pie',autopct = '%.2f')
```

a. Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'])
```

b. Distplot

```python
sns.distplot(df['Age'])
```

c. Boxplot

```python
sns.boxplot(df['Age'])
```
