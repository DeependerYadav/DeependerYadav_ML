# Scatterplot(Numerical - Numerical)

```python
# 1. On X- axis we have mpg(miles per gallon of a car)
# 2. On Y- axis we have the weight of car
# 3. hue = using satuartion of a point we can find how many cylinders a car have
# 4. style = Using this we can change the style of a point
# 5. size = Change the size of a point according to data
```

```python
sns.scatterplot(x=mpg['mpg'], y=mpg['weight'], hue = mpg['cylinders'], style = mpg['origin'], size = mpg['horsepower'])
```

# Barplot(Numerical - Categorical)

```python
titanic.head()
```

```python
sns.barplot(x = titanic['Pclass'],y = titanic['Age'],hue=titanic['Sex'])
```

# Boxplot(Numerical - Categorical)

```python
sns.boxplot(x = titanic['Sex'],y = titanic['Age'],hue=titanic['Survived'])
```

# Distplot (Numerical - Categorical)

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist=False)
```

# HeatMap (Categorical - Categorical)
